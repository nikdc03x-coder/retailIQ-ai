# 01 — Walmart Dataset Exploration
Understand the data before cleaning or modeling.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Reuse the installed project configuration.
from retailiq.config import ROOT as root

# Load the original datasets.
sales = pd.read_csv(root / "data/raw/train.csv")
features = pd.read_csv(root / "data/raw/features.csv")
stores = pd.read_csv(root / "data/raw/stores.csv")

print("Sales:", sales.shape)
print("Features:", features.shape)
print("Stores:", stores.shape)
```

## 1. Inspect the dataset
Check column names, sample records and data types.

```python
# Preview each dataset.
display(sales.head())
display(features.head())
display(stores.head())

# Check the sales column types.
print(sales.dtypes)
```

## 2. Check the historical period

```python
# Inspect the dates in the sales records.
dates = pd.to_datetime(sales["Date"], errors="coerce")
print("First date:", dates.min())
print("Last date:", dates.max())
print("Unique dates:", dates.nunique())
print("Invalid dates:", dates.isna().sum())
```

## 3. Visualize sales
This chart shows how sales change across weeks.

```python
# Combine sales for each recorded week.
weekly = sales.assign(Date=dates).groupby("Date")["Weekly_Sales"].sum()

# Plot historical sales over time.
fig, ax = plt.subplots(figsize=(11, 4))
weekly.plot(ax=ax)
ax.set_title("Total Weekly Sales")
ax.set_xlabel("Date")
ax.set_ylabel("Weekly Sales")
plt.show()
```
